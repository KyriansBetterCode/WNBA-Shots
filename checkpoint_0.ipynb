```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('/Users/grindhardt/Desktop/Python/Project STAT/wnba_shots_2025.csv')
print("Dataset Shape:", df.shape)
df[['PLAYER_ID', 'PLAYER_NAME', 'ACTION_TYPE', 'SHOT_DISTANCE', 'LOC_X', 'LOC_Y', 'SHOT_MADE_FLAG']].head()
```

```python
# 1. Feature flag for moving/off-dribble shot mechanics vs set shots
df['IS_MOVING_SHOT'] = df['ACTION_TYPE'].str.contains(
    'Pullup|Step Back|Fadeaway|Floating|Driving|Running|Cut|Cutting', regex=True, case=False
).astype(int)

# 2. Derive exact continuous Euclidean distance from raw LOC_X and LOC_Y 
df['CALC_DIST_FT'] = np.sqrt(df['LOC_X']**2 + df['LOC_Y']**2) / 10.0

df[['ACTION_TYPE', 'IS_MOVING_SHOT', 'SHOT_DISTANCE', 'CALC_DIST_FT']].head()
```
