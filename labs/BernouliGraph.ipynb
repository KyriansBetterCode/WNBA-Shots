```python
import matplotlib.pyplot as plt
import pandas as pd

# Load your dataset (update the filename if needed)
df_shots = pd.read_csv('wnba_shots_2025.csv')

# Calculate proportions and total count
proportions = df_shots['SHOT_MADE_FLAG'].value_counts(normalize=True)
n_total = len(df_shots)

# Set up labels and values in correct order (0 = Missed, 1 = Made)
labels = ['Missed (0)', 'Made (1)']
values = [proportions.get(0, 0), proportions.get(1, 1)]

# Create the figure
plt.figure(figsize=(6, 5))
bars = plt.bar(
    labels,
    values,
    color=['#6b7c93', '#d9532d'],  # Muted grey-blue and burnt orange
    width=0.45,
)

# Styling the plot to match your reference image
plt.title(
    f'WNBA shots: distribution of the outcome (n = {n_total:,})',
    fontweight='bold',
    fontsize=12,
    pad=12,
)
plt.ylabel('Proportion of shots', fontsize=11)
plt.ylim(0, 0.7)  # Match y-axis limit from 0 to 0.7
plt.grid(axis='y', linestyle='-', alpha=0.3)

# Add the exact proportion labels on top of each bar
for bar in bars:
  height = bar.get_height()
  plt.text(
      bar.get_x() + bar.get_width() / 2.0,
      height + 0.015,
      f'{height:.3f}',
      ha='center',
      va='bottom',
      fontweight='bold',
      fontsize=11,
      color='#222222',
  )

# Remove top and right spines for a clean look
sns_ax = plt.gca()
sns_ax.spines['top'].set_visible(False)
sns_ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```
