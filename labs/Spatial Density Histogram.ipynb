```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
plt.figure(figsize=(6, 5))
sns.histplot(data=pd.read_csv('wnba_shots_2025.csv'), x='LOC_X', y='LOC_Y', bins=30, cmap='YlOrRd', cbar=True)
plt.title('2D Spatial Density Histogram of WNBA Shot Attempts')
plt.xlabel('LOC_X (10ths of a foot)')
plt.ylabel('LOC_Y (10ths of a foot)')
plt.savefig(
    'wnba_spatial_density_histogram.png', dpi=300, bbox_inches='tight'
)
plt.show()
```
