```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
rng = np.random.default_rng(651)
```

```python
df = pd.read_csv('/Users/grindhardt/Desktop/Python/Project STAT/wnba_shots_2025.csv')
print(df.shape)
print(df.head(10))
print(df.isna().sum().sort_values(ascending=False).head(10))
```

```python
OUTCOME = 'SHOT_MADE_FLAG'
GROUP = 'ACTION_TYPE' 
ID_COLS = ['GAME_ID', 'PLAYER_ID', 'TEAM_ID', 'GAME_EVENT_ID']
y = (df[OUTCOME] == 1).to_numpy().astype(int) # a 0/1 outcome
# numeric outcome instead: y = df[OUTCOME].to_numpy
```

#### E2 (The Outcome)

```python
# Convert outcome to 0/1 array and compute overall rate
y = (df[OUTCOME] == 1).to_numpy().astype(int)
print("Overall Field Goal Rate (Yes/No):", y.mean().round(4))

# Plotting the discrete outcome
plt.hist(y, bins=[-0.5, 0.5, 1.5], rwidth=0.8)
plt.xticks([0, 1], ['Miss (0)', 'Make (1)'])
plt.title('Distribution of Shot Outcomes (Bernoulli Trial)')
plt.show()

# Something that would show I am wrong would be finding a probabality of a shot being made that is less than 0 or greater than 1, which is impossible for a Bernoulli trial.
```

```python
# E3: correlations and PCA on numeric inputs only
num = df.select_dtypes("number").drop(columns=ID_COLS + [OUTCOME], errors='ignore')
num = num.loc[:, num.std() > 0]  # Drop constant columns
R = num.corr()

pairs = [
    (a, b, round(float(R.loc[a, b]), 3)) 
    for i, a in enumerate(R.columns) 
    for b in R.columns[i + 1:] 
    if abs(R.loc[a, b]) > 0.8
]

print(pairs)
Z = ((num - num.mean()) / num.std()).fillna(0)  #standardize; gap -> mean
S = np.linalg.svd(Z.to_numpy(), compute_uv=False)
cum = np.cumsum(S**2) / (S**2).sum()
print(np.argmax(cum >= 0.90) + 1, "components reach 90%")
```

#### E4 (Your one number)

```python
p_hat = y.mean()
n = len(y)

# Fisher SE & Wald 95% CI
se_fisher = np.sqrt(p_hat * (1 - p_hat) / n)
ci_fisher = (p_hat - 1.96 * se_fisher, p_hat + 1.96 * se_fisher)

# Bootstrap 95% CI (10,000 resamples)
boot = [rng.choice(y, len(y), replace=True).mean() for _ in range(10_000)]
ci_boot = np.percentile(boot, [2.5, 97.5])

print(f"MLE Estimate: {p_hat:.4f}")
print(f"Fisher 95% CI: {np.round(ci_fisher, 4)}")
print(f"Bootstrap 95% CI: {np.round(ci_boot, 4)}")
```

```python
# E5: the one number in each subgroup, with a bootstrap interval
for level in df[GROUP].dropna().unique():
    ys = y[(df[GROUP] == level).to_numpy()]
    boot = [rng.choice(ys, len(ys), replace=True).mean() for _ in range(10_000)]
    print(level, len(ys), ys.mean().round(4),
            np.percentile(boot, [2.5, 97.5]).round(4))
```
